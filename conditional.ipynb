```python
age = int(input("Enter your age:"))
if age<=3:
    print("child")
elif age>=4 and age<=12:
    print("kid")
elif age>=13 and age<=18:
    print("teenager")
elif age>=19 and age<=60:
    print("adult")
else:
    print("thala")
```
