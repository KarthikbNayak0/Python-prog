Iteration over dict

```python
student={
    'name': 'rajshekar',
    'id': 'cs000',
    'age':'23'
}

for x in student.values():
    print(x)
print("\n")
for k in student:
    print(k)
print("\n")
for i,j in student.items():
    print(i," ",j)
```
