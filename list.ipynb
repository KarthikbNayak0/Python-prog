```python
fruits=["apple","banana","mango","grapes"]
print(fruits)
```

```python
#indexing
print(fruits[2])
print(fruits[-1])#last element using index
#slicing
print(fruits[1:3])
rev_fruits=fruits[::-1]
print(rev_fruits)
```

```python
#modifing list
print("orginal list:",fruits)
#add the element at the end of the list
fruits.append("kiwi")
print("after adding element at the end:",fruits)
#to findlen of the list
list_length=len(fruits)
print("length of the list:",list_length)
#to remove the last element
fruits.pop()
print("after removing last element:",fruits)
# remove the element
'''rem_fruits=fruits.remove("mango")
print(fruits)'''
#replace the element
fruits[3]="watermelon"
print(fruits)
```

```python
nums=[1,42,34,4,5]
print(max(nums))
print(min (nums))
nums.sort()
print(nums)
```

```python
#looping
students=["karthik","sachin","rohit","virat"]
for i in range(len(students)):
    print(students[i])
print("\n")
for st in students:
    print(st)
for val in students:
    print(val,end=" ")
```

```python
nums=[1,2,3,4,5]
colors=["red","green","blue"]
list3=nums+colors
print(list3)
nums.extend(colors)
print(nums)
```

```python
n=5
sqr=[i*i for i in range(1,n+1)]
print(sqr)
```

```python
n=10
even=[i for i in range(1,n+1) if i%2==0]
print(even)

odd=[i for i in range(1,n+1) if i%2!=0]
print(odd)
```
