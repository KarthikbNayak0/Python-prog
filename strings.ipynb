Indexing and slicing

```python
str="darkchocolate"
```

```python
#indexing
print(str[5])
print(str[-len(str)])#last element using index
print(str[-3])#element from last using negative index
```

```python
#slicing
print(str[0:4])#st_idx=0, end_idx=4, step=1(d)
print(str[4:])#st_idx=4, end_idx=len(str), step=1(d)
print(str[len(str)-1::-1])
#to  reverse a string
rev_str=str[::-1]
print(rev_str)
```

String Function

```python
s="he  is a coder"
```

```python
print(s+" and he is a good coder")
print(s)
#length of string
length=len(s)
print(length)
#end with substring
print(s.endswith("coder"))
#replace old,new
print(s.replace("coder","programmer"))
#capitalize
print(s.capitalize())
#find fisrt index of substring
print(s.find("s"))
#count of substring
print(s.count("e"))
#upper lower
print(s.upper())
print(s.lower())
```
