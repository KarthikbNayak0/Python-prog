Python Tuples

a tuple is aa collection of items in python that is ordered,unchangeable(immutable) and allow duplicate values.

```python
#-----------types-----------
#type 1:using parenthesis
nums=(1,2,3,4,5)
mixed=("raju",23,43.56,True)
nested=(21,(23,54),(67,68,69))
print(nums,type(nums))
#type 2:without using parenthesis
colors="red","green","blue"
print(colors,type(colors))
#type 3:using tuple const
list=['a','b','c','d']
tuple_list=tuple(list)
print(tuple_list,type(tuple_list))
#type4:single
value=(10,)
print(value,type(value))
```

Tuple operation


```python

#concatenation 
tup1=(1,2,3)
tup2=(4,5,6)
tuple3=tup1+tup2
print(tuple3)

st_tup2=list(tup2)
print(st_tup2,type(st_tup2))
st_tup2[1]=100
print(st_tup2,type(st_tup2))
tup2=tuple(st_tup2)
print(tup2,type(tup2))
```

```python
marks=(40,)*5
print(marks)
tup=(10,20,30,40)
print(50 in tup)
```

```python
colors=("red","green","blue","yellow")
for c in colors:
    print(c,end=" ")
print("\n")
i=0
while i<len(colors):
    print(colors[i],end=" ")
    i+=1
```

```python
colors=("red","green","blue","yellow","red","green")
print(colors.count("red"))
print(colors.index("red"))
```
